```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.data import Data, DataLoader, InMemoryDataset
from torch_geometric.nn import GCNConv, global_mean_pool
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("cardio_train.csv", sep=';')  
df = df.drop(columns=['id'])
df['age'] = (df['age'] / 365).astype(int)

X = df.drop(columns=['cardio'])
y = df['cardio'].values

X_train_df, X_test_df, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train_df.values)
X_test = scaler.transform(X_test_df.values)

feature_names = X.columns.tolist()
num_nodes = X_train.shape[1] 

corr = np.corrcoef(X_train, rowvar=False)  
thr = 0.8  # correlation threshold for creating edges
adj = (np.abs(corr) >= thr).astype(int)
np.fill_diagonal(adj, 0)

sources, targets = np.where(adj == 1)
edge_index = torch.tensor(np.vstack((sources, targets)), dtype=torch.long)

edge_index = torch.cat([edge_index, edge_index[[1,0]]], dim=1)

edge_index = torch.unique(edge_index, dim=1)

from torch_geometric.data import Data

# from torch_geometric.data import Data

def make_data_array(X_arr, y_arr):
    data_list = []
    for i in range(X_arr.shape[0]):
        node_feats = torch.tensor(X_arr[i].reshape(num_nodes, 1), dtype=torch.float)
        label = torch.tensor([y_arr[i]], dtype=torch.float)  # ✅ use direct indexing
        data = Data(x=node_feats, edge_index=edge_index, y=label)
        data_list.append(data)
    return data_list




train_data = make_data_array(X_train, y_train)
test_data  = make_data_array(X_test, y_test)

train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader  = DataLoader(test_data, batch_size=64, shuffle=False)

class SBGC_LSTM(nn.Module):
    def __init__(self, node_feat_dim=1, gcn_hidden=32, lstm_hidden=64, num_layers_gcn=2, num_classes=2):
        super().__init__()
        self.gcn_layers = nn.ModuleList()
        in_ch = node_feat_dim
        for _ in range(num_layers_gcn):
            self.gcn_layers.append(GCNConv(in_ch, gcn_hidden))
            in_ch = gcn_hidden
        self.lstm = nn.LSTM(input_size=gcn_hidden, hidden_size=lstm_hidden,
                            num_layers=1, batch_first=True, bidirectional=True)
        self.classifier = nn.Sequential(
            nn.Linear(2 * lstm_hidden, 64), 
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(64, num_classes)
        )
    def forward(self, data):
        x, edge_index, batch = data.x, data.edge_index, data.batch
        for conv in self.gcn_layers:
            x = conv(x, edge_index)
            x = F.relu(x)
        batch_size = batch.max().item() + 1
        num_nodes_per_graph = num_nodes  
        gcn_h = x.view(batch_size, num_nodes_per_graph, -1)  
        lstm_out, (h_n, c_n) = self.lstm(gcn_h)  
        graph_emb = torch.mean(lstm_out, dim=1) 
        logits = self.classifier(graph_emb)  
        return logits
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = SBGC_LSTM(node_feat_dim=1, gcn_hidden=64, lstm_hidden=64, num_layers_gcn=2, num_classes=2).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-4)
criterion = nn.CrossEntropyLoss()
# criterion = nn.CrossEntropyLoss()
def train_epoch():
    model.train()
    total_loss = 0
    correct = 0
    total = 0
    for batch in train_loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        out = model(batch)               
        loss = criterion(out, batch.y.view(-1).long())
        loss.backward()
        optimizer.step()
        total_loss += loss.item() * batch.num_graphs
        preds = out.argmax(dim=1)
        correct += (preds == batch.y.view(-1)).sum().item()
        total += batch.num_graphs
    return total_loss / total, correct / total

def evaluate(loader):
    model.eval()
    correct = 0
    total = 0
    all_labels = []
    all_preds = []
    with torch.no_grad():
        for batch in loader:
            batch = batch.to(device)
            out = model(batch)
            preds = out.argmax(dim=1)
            correct += (preds == batch.y.view(-1)).sum().item()
            total += batch.num_graphs
            all_labels.extend(batch.y.view(-1).cpu().numpy().tolist())
            all_preds.extend(preds.cpu().numpy().tolist())
    acc = correct / total
    return acc, all_labels, all_preds
n_epochs = 25
for epoch in range(1, n_epochs+1):
    train_loss, train_acc = train_epoch()
    test_acc, _, _ = evaluate(test_loader)
    print(f"Epoch {epoch:02d} | Train Loss: {train_loss:.4f} | Train Acc: {train_acc:.4f} | Test Acc: {test_acc:.4f}")
```
