```python
from scipy.stats import shapiro

for feature in numeric_features:
    stat, p = shapiro(df[feature])
    print(f'{feature} -> Statistics={stat:.3f}, p={p:.3f}')
    alpha = 0.05
    if p > alpha:
        print('Sample looks Gaussian (fail to reject H0)\n')
    else:
        print('Sample does not look Gaussian (reject H0)\n')
```

```python
from scipy.stats import chi2_contingency

for feature in categorical_features:
    stat, p, dof, expected = chi2_contingency(
        pd.crosstab(df[feature], df['HeartDisease'])
    )

    print(f"{feature} -> p value = {p}")
    alpha = 0.05
    if p <= alpha:
        print('Dependent (reject H0)\n')
    else:
        print('Independent (H0 holds true)\n')
```

```python
y = train['HeartDisease']
X = train.drop('HeartDisease', axis=1)
print(X.shape, y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.1, random_state=42, stratify=y
)
print("Train size:", X_train.shape)
print("Test size:", X_test.shape)
```

```python
from xgboost import XGBClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Define only XGBoost model
xgb_model = XGBClassifier(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=5,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric='logloss',
    random_state=42
)

# Train the model
xgb_model.fit(X_train, y_train)

# Predict
y_pred = xgb_model.predict(X_test)

# Evaluation
print("\nModel: XGBoost")
print(f'Accuracy: {accuracy_score(y_test, y_pred)}')
print(f'Precision: {precision_score(y_test, y_pred)}')
print(f'Recall: {recall_score(y_test, y_pred)}')
print(f'F1-score: {f1_score(y_test, y_pred)}')
print('-'*40)
```

```python
# Run this in Jupyter — generates XGBoost result graphs
import matplotlib.pyplot as plt
import numpy as np
from sklearn.metrics import confusion_matrix, roc_curve, roc_auc_score
import seaborn as sns
import joblib
import pandas as pd
from xgboost import XGBClassifier

# Load your saved model and test data
# (adjust paths to wherever your files are)
xgb_model = XGBClassifier()
xgb_model.load_model("xgboost_heart_model.json")

# You need your test X and y — load from wherever you saved them
# If you saved them during training, load here:
# X_test = pd.read_csv("X_test.csv")
# y_test = pd.read_csv("y_test.csv")

# Once you have X_test and y_test:
y_pred_prob = xgb_model.predict_proba(X_test)[:, 1]
y_pred      = (y_pred_prob >= 0.3).astype(int)

# Plot 1 — Confusion Matrix + ROC
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

cm = confusion_matrix(y_test, y_pred)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", ax=axes[0],
            xticklabels=["No Disease","Heart Disease"],
            yticklabels=["No Disease","Heart Disease"])
axes[0].set_xlabel("Predicted"); axes[0].set_ylabel("Actual")
axes[0].set_title("Confusion Matrix — Clinical Model")

fpr, tpr, _ = roc_curve(y_test, y_pred_prob)
auc = roc_auc_score(y_test, y_pred_prob)
axes[1].plot(fpr, tpr, color="blue", lw=2, label=f"AUC={auc:.4f}")
axes[1].plot([0,1],[0,1],"k--",lw=1)
axes[1].set_xlabel("False Positive Rate")
axes[1].set_ylabel("True Positive Rate")
axes[1].set_title("ROC Curve — Clinical Model")
axes[1].legend(loc="lower right")
axes[1].grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig("xgb_evaluation.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: xgb_evaluation.png")

# Plot 2 — Feature Importance
fig, ax = plt.subplots(figsize=(8, 5))
features = ["BMI","Smoking","Stroke","PhysicalHealth","DiffWalking",
            "Sex","AgeCategory","Diabetic","PhysicalActivity","SleepTime","KidneyDisease"]
importances = xgb_model.feature_importances_
idx = np.argsort(importances)
ax.barh([features[i] for i in idx], importances[idx], color="#2563EB")
ax.set_xlabel("Feature Importance Score")
ax.set_title("XGBoost — Feature Importance")
ax.grid(True, alpha=0.3, axis="x")
plt.tight_layout()
plt.savefig("xgb_feature_importance.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved: xgb_feature_importance.png")
```
